```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Load the cleaned dataset from Task 1
df = pd.read_csv('cleaned_netflix_data.csv')
print("Cleaned dataset loaded successfully! Total rows:", len(df))

# Step 2: Calculate total counts and proportions of Movies and TV Shows
type_counts = df['type'].value_counts()
type_percentages = df['type'].value_counts(normalize=True) * 100

print("\n--- Content Type Counts ---")
print(type_counts)

print("\n--- Content Type Percentages (%) ---")
print(type_percentages.round(2))

# Step 3: Create visualizations for content distribution
sns.set_theme(style="whitegrid")
plt.figure(figsize=(7, 5))

# Create countplot comparing Movies vs TV Shows
ax = sns.countplot(data=df, x='type', palette=['#E50914', '#221F1F'])

plt.title('Distribution of Movies vs TV Shows on Netflix', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Content Type', fontsize=12)
plt.ylabel('Total Count', fontsize=12)

# Add exact count labels on top of the bars
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='center', 
                xytext=(0, 5), 
                textcoords='offset points',
                fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()
```
