```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Load the cleaned dataset from Task 1
df = pd.read_csv('cleaned_netflix_data.csv')
print("Cleaned dataset loaded successfully! Total rows:", len(df))

# Step 2: Filter for recent release years (1990 to present) for better trend visibility
trend_df = df[df['release_year'] >= 1990]

# Step 3: Group by release year and content type
year_type_counts = trend_df.groupby(['release_year', 'type']).size().unstack(fill_value=0)

print("\n--- Content Released Per Year (Recent 10 Years) ---")
print(year_type_counts.tail(10))

# Step 4: Create trend line chart visualization
sns.set_theme(style="whitegrid")
plt.figure(figsize=(11, 6))

plt.plot(year_type_counts.index, year_type_counts['Movie'], label='Movies', color='#E50914', linewidth=2.5, marker='o')
plt.plot(year_type_counts.index, year_type_counts['TV Show'], label='TV Shows', color='#221F1F', linewidth=2.5, marker='s')

plt.title('Netflix Content Addition Trends by Release Year (1990 - Present)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Release Year', fontsize=12)
plt.ylabel('Number of Titles Released', fontsize=12)
plt.legend(title='Content Type', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
