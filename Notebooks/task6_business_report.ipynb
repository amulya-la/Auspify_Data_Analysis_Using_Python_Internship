```python
import pandas as pd
import numpy as np

# Step 1: Load the cleaned dataset from Task 1
df = pd.read_csv('cleaned_netflix_data.csv')
print("Cleaned dataset loaded successfully! Total catalog rows analyzed:", len(df))

# Step 2: Advanced Metric Calculations (using safe string matching)
total_titles = len(df)
movie_count = len(df[df['type'].str.lower().str.contains('movie', na=False)])
tv_count = len(df[df['type'].str.lower().str.contains('tv show', na=False)])

# Fallback if counts are split differently
if tv_count == 0:
    tv_count = total_titles - movie_count

# Top Content Producing Country
top_country = df[df['country'] != 'Unknown']['country'].str.split(', ').explode().mode()[0]

# Top Director (excluding 'Unknown' / 'Not Given')
valid_directors = df[~df['director'].isin(['Unknown', 'Not Given', 'nan'])]
top_director = valid_directors['director'].mode()[0] if not valid_directors.empty else 'Not Given'

# Most Common Rating
top_rating = df['rating'].mode()[0]

# Peak Release Year
peak_year = df['release_year'].mode()[0]

# Step 3: Print Formal Executive Business Insights Report
print("\n" + "="*60)
print("             NETFLIX STRATEGIC BUSINESS INSIGHTS REPORT")
print("             Auspify Data Analysis Internship - Task 6")
print("="*60)

print(f"\n[1] EXECUTIVE PORTFOLIO OVERVIEW")
print(f" • Total Catalog Volume: {total_titles} verified titles")
print(f" • Content Breakdown: {movie_count} Movies ({round(movie_count/total_titles*100, 2)}%) | {tv_count} TV Shows ({round(tv_count/total_titles*100, 2)}%)")

print(f"\n[2] GEOGRAPHIC & PRODUCTION LEADERSHIP")
print(f" • Primary Production Hub: {top_country} leads global contributions.")
print(f" • Peak Production Vintage: Year {peak_year} witnessed the highest volume of content releases.")

print(f"\n[3] CATALOGUE GOVERNANCE & CREATIVE INSIGHTS")
print(f" • Most Proactive Director: {top_director}")
print(f" • Dominant Maturity Rating: {top_rating} represents the largest share of audience classification.")

print(f"\n[4] STRATEGIC RECOMMENDATIONS FOR PORTFOLIO GROWTH")
print(f" • Recommendation A: Expand regional content acquisition outside of primary hubs to capture emerging international markets.")
print(f" • Recommendation B: Balance episodic series production against feature films to improve user retention rates.")
print(f" • Recommendation C: Target family-oriented and teen ratings categories to diversify subscriber demographics.")

print("\n" + "="*60)
print(" TASK 6 ADVANCED REPORT GENERATED SUCCESSFULLY.")
print("="*60)
```
