```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Load the cleaned dataset from Task 1
df = pd.read_csv('cleaned_netflix_data.csv')
print("Cleaned dataset loaded successfully! Total rows:", len(df))

# Step 2: Extract top ratings distribution
rating_counts = df['rating'].value_counts().head(10)

# Step 3: Extract top genres (exploding comma-separated listed_in column)
genre_series = df['listed_in'].str.split(', ').explode()
top_genres = genre_series.value_counts().head(10)

# Step 4: Create a 2-in-1 Dashboard Visualization
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Subplot 1: Top 10 Ratings
sns.barplot(ax=axes[0], x=rating_counts.values, y=rating_counts.index, palette='Blues_r')
axes[0].set_title('Top 10 Ratings Distribution on Netflix', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Total Titles', fontsize=11)
axes[0].set_ylabel('Rating', fontsize=11)

for i, count in enumerate(rating_counts.values):
    axes[0].text(count + 20, i, str(count), va='center', fontsize=9, fontweight='bold')

# Subplot 2: Top 10 Genres
sns.barplot(ax=axes[1], x=top_genres.values, y=top_genres.index, palette='Purples_r')
axes[1].set_title('Top 10 Genres on Netflix', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Total Titles', fontsize=11)
axes[1].set_ylabel('Genre / Category', fontsize=11)

for i, count in enumerate(top_genres.values):
    axes[1].text(count + 20, i, str(count), va='center', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()
```
